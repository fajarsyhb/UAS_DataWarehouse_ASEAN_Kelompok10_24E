```python
import pandas as pd
import numpy as np
import atoti as tt
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('clean/worldbank_clean.csv')

print(f"Shape data: {df.shape}")
print(f"Kolom: {list(df.columns)}")
print(f"Tahun: {sorted(df['year'].unique())}")
print(f"Negara: {sorted(df['economy'].unique())}")
df.head(10)
```

```python
country_map = {
    'IDN': 'Indonesia', 'MYS': 'Malaysia', 'THA': 'Thailand',
    'SGP': 'Singapore', 'PHL': 'Philippines', 'VNM': 'Vietnam',
    'MMR': 'Myanmar', 'KHM': 'Cambodia', 'LAO': 'Laos', 'BRN': 'Brunei'
}

df['country_name']  = df['economy'].map(country_map)
df['region']        = 'ASEAN'
df['year']          = df['year'].astype(int)
df['gdp_usd']           = pd.to_numeric(df['gdp_usd'], errors='coerce')
df['inflation_pct']     = pd.to_numeric(df['inflation_pct'], errors='coerce')
df['unemployment_pct']  = pd.to_numeric(df['unemployment_pct'], errors='coerce')
df['population']        = pd.to_numeric(df['population'], errors='coerce')
df['gdp_billion_usd']   = df['gdp_usd'] / 1e9

print(df.dtypes)
print(f"\nMissing values:\n{df.isnull().sum()}")
df.head()
```

Buat Session & Load data

```python
session = tt.Session.start()
table = session.read_pandas(df, table_name="EconomyASEAN")
print("Session & table berhasil dibuat ✓")
table.head()
```

```python
cube = session.create_cube(table)
print("Cube berhasil dibuat ✓")
print(cube)
```

```python
m = cube.measures
l = cube.levels

print("=== MEASURES ===")
for name in m:
    print(" -", name)

print("\n=== LEVELS ===")
for name in l:
    print(" -", name)
```

Tambah Custom Measures

```python
m["GDP Avg (USD)"]            = tt.agg.mean(table["gdp_usd"])
m["GDP (Billion USD)"]        = tt.agg.mean(table["gdp_billion_usd"])
m["Inflasi Rata-rata (%)"]    = tt.agg.mean(table["inflation_pct"])
m["Pengangguran Rata-rata (%)"] = tt.agg.mean(table["unemployment_pct"])
m["Populasi Total"]           = tt.agg.sum(table["population"])

print("Custom measures berhasil ✓")
```

```python
cube.query(
    m["GDP (Billion USD)"],
    m["Inflasi Rata-rata (%)"],
    m["Pengangguran Rata-rata (%)"],
    levels=[l["country_name"], l["period"]]
)
```

```python
cube.query(
    m["GDP (Billion USD)"],
    m["Populasi Total"],
    levels=[l["country_name"]],
).sort_values("GDP (Billion USD)", ascending=False)
```

```python
cube.query(
    m["Inflasi Rata-rata (%)"],
    m["Pengangguran Rata-rata (%)"],
    levels=[l["region"], l["period"]]
)
```

Hasil dan Analisis Hasil 

Analisis 1: GDP per negara per periode

```python
hasil_gdp = cube.query(
    m["GDP (Billion USD)"],
    m["Inflasi Rata-rata (%)"],
    m["Pengangguran Rata-rata (%)"],
    levels=[l["country_name"], l["period"]]
).sort_values("GDP (Billion USD)", ascending=False)

print("=== RANKING GDP PER NEGARA PER PERIODE ===")
hasil_gdp
```

Analisis 2: Ranking GDP tertinggi

```python
ranking_gdp = cube.query(
    m["GDP (Billion USD)"],
    m["Populasi Total"],
    levels=[l["country_name"]]
).sort_values("GDP (Billion USD)", ascending=False)

print("=== RANKING GDP TERTINGGI ASEAN ===")
ranking_gdp
```

Analisis 3: Negara inflasi tertinggi

```python
ranking_inflasi = cube.query(
    m["Inflasi Rata-rata (%)"],
    m["GDP (Billion USD)"],
    levels=[l["country_name"]]
).sort_values("Inflasi Rata-rata (%)", ascending=False)

print("=== RANKING INFLASI TERTINGGI ASEAN ===")
ranking_inflasi
```

Analisis 4: Pengangguran tertinggi

```python
ranking_nganggur = cube.query(
    m["Pengangguran Rata-rata (%)"],
    m["Populasi Total"],
    levels=[l["country_name"]]
).sort_values("Pengangguran Rata-rata (%)", ascending=False)

print("=== RANKING PENGANGGURAN TERTINGGI ASEAN ===")
ranking_nganggur
```
